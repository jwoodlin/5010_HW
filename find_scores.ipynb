```python
games = pd.read_csv(
    "/projects/class/itcs5010_u01/SportsTrackingTransformer/data/BigDataBowl_2024/games.csv"
)

games.head()
```

```python
original_game = games[
    (games["week"] == 6) &
    (
        ((games["homeTeamAbbr"] == "LAC") & (games["visitorTeamAbbr"] == "DEN")) |
        ((games["homeTeamAbbr"] == "DEN") & (games["visitorTeamAbbr"] == "LAC"))
    )
]

print(original_game[
    ["gameId", "week", "homeTeamAbbr", "visitorTeamAbbr",
     "homeFinalScore", "visitorFinalScore"]
].to_string(index=False))
```

```python
games2 = pd.read_csv(
    "/projects/class/itcs5010_u01/SportsTrackingTransformer/data/BigDataBowl_2024/games2.csv"
)

modified_game = games2[
    (games2["week"] == 6) &
    (
        ((games2["homeTeamAbbr"] == "LAC") & (games2["visitorTeamAbbr"] == "DEN")) |
        ((games2["homeTeamAbbr"] == "DEN") & (games2["visitorTeamAbbr"] == "LAC"))
    )
]

print(modified_game[
    ["gameId", "week", "homeTeamAbbr", "visitorTeamAbbr",
     "homeFinalScore", "visitorFinalScore"]
].to_string(index=False))
```
